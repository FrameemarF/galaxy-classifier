```python
import sys
from pathlib import Path

project_root = Path.cwd().parent
sys.path.append(str(project_root))

import torch.nn as nn
import config
import torch
from torch.optim import Adam
from src.model import SimpleGalaxyCNN
from src.train import train_one_epoch, evaluate
from src.dataset import create_dataloaders
from src.transforms import preprocess_image


model = SimpleGalaxyCNN()
criterion = nn.CrossEntropyLoss()
optimizer = Adam(model.parameters(), lr=0.001)

num_epochs = 13
train_loader, val_loader, test_loader = create_dataloaders(
    config.PROCESSED_H5_PATH, config.SPLIT_INDICES_PATH, preprocess_image, batch_size=32
)

best_val_acc = 0.0
for epoch in range(num_epochs):
    train_loss, train_acc = train_one_epoch(model, train_loader, optimizer, criterion)
    val_loss, val_acc = evaluate(model, val_loader, criterion)
    print(f"Epoch {epoch+1}/{num_epochs}: train_loss={train_loss:.4f}, train_acc={train_acc:.4f} val_loss={val_loss:.4f}, val_acc={val_acc:.4f}")
    if (val_acc > best_val_acc):
            torch.save(model.state_dict(), config.MODELS_DIR / "best_model.pt")
            best_val_acc = val_acc
            print(f"===best model saved, epoch: {epoch+1}")
```
