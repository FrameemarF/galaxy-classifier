```python
import sys
from pathlib import Path

project_root = Path.cwd().parent
sys.path.append(str(project_root))

import config
import json
import numpy as np
from src.dataset import GalaxyDataset
from src.transforms import preprocess_image
from torch.utils.data import DataLoader
from src.model import SimpleGalaxyCNN


with open(config.SPLIT_INDICES_PATH, 'r') as f:
    split_data = json.load(f)

train_idx = np.array(split_data['train'])
val_idx = np.array(split_data['val'])
test_idx = np.array(split_data['test'])

train_dataset = GalaxyDataset(config.RAW_H5_PATH, train_idx, transform=preprocess_image)
val_dataset = GalaxyDataset(config.RAW_H5_PATH, val_idx, transform=preprocess_image)
test_dataset = GalaxyDataset(config.RAW_H5_PATH, test_idx, transform=preprocess_image)

train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=32, shuffle=False)

model = SimpleGalaxyCNN()
images, labels = next(iter(train_loader))
output = model(images)
print(output.shape)
```
